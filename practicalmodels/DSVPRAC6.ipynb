```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create airline flights dataset directly
data = {
    "carrier": [
        "AA", "AA", "AA", "AA", "AA",
        "DL", "DL", "DL", "DL", "DL",
        "UA", "UA", "UA", "UA", "UA",
        "B6", "B6", "B6", "B6", "B6",
        "WN", "WN", "WN", "WN", "WN",
        "EV", "EV", "EV", "EV", "EV"
    ],

    "dep_delay": [
        5, 15, -3, 30, 45,
        2, 10, -5, 25, 60,
        0, 20, 35, -2, 50,
        8, 18, -4, 40, 55,
        -1, 12, 28, 5, 70,
        3, 22, -6, 35, 80
    ],

    "distance": [
        2475, 184, 1096, 733, 2475,
        214, 762, 1089, 1399, 2586,
        1400, 719, 1096, 2475, 1440,
        1065, 1521, 1035, 1598, 2296,
        761, 1035, 1598, 838, 2586,
        533, 888, 1096, 1504, 2454
    ]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Display dataset
print(df)
```

```python
# Create delay status
df["delay_status"] = df["dep_delay"].apply(
    lambda x: "On-time" if x <= 0 else "Delayed"
)

print("\nDataset with Delay Status:")
print(df)
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="carrier",
    y="dep_delay",
    hue="delay_status"
)

plt.title("Distribution of Departure Delay by Airline Carrier")
plt.xlabel("Airline Carrier")
plt.ylabel("Departure Delay (minutes)")
plt.legend(title="Delay Status")

plt.show()
```
