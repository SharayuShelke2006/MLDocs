```python
# 1. Import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


# 2. Load the dataset
df = pd.read_csv("synthetic_airline_flights.csv")
```

Average Arrival Delay

```python
avg_delay = df.groupby("Airline")["Arrival_Delay"].mean()

print("\nAverage Arrival Delay by Airline:")
print(avg_delay)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_delay.index,
    y=avg_delay.values
)

plt.title("Average Arrival Delay by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Arrival Delay (minutes)")

plt.show()
```

Departure Delay By Airline

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Airline",
    y="Departure_Delay"
)

plt.title("Departure Delay by Airline")
plt.xlabel("Airline")
plt.ylabel("Departure Delay (minutes)")

plt.show()
```

Arrival Delay By airline

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Airline",
    y="Arrival_Delay"
)

plt.title("Arrival Delay by Airline")
plt.xlabel("Airline")
plt.ylabel("Arrival Delay (minutes)")

plt.show()
```

Flight Distance vs Arrival Delay

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Flight_Distance",
    y="Arrival_Delay"
)

plt.title("Flight Distance vs Arrival Delay")
plt.xlabel("Flight Distance (miles)")
plt.ylabel("Arrival Delay (minutes)")

plt.show()
```

Departure Delay vs Arrival Delay

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Departure_Delay",
    y="Arrival_Delay"
)

plt.title("Departure Delay vs Arrival Delay")
plt.xlabel("Departure Delay (minutes)")
plt.ylabel("Arrival Delay (minutes)")

plt.show()
```

```python
# ------------------------------------------
# 15. Distribution of Arrival Delays
# ------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Arrival_Delay",
    bins=30,
    kde=True
)

plt.title("Distribution of Arrival Delays")
plt.xlabel("Arrival Delay (minutes)")
plt.ylabel("Number of Flights")

plt.show()
```

```python
# ------------------------------------------
# 16. Regression Plot
# ------------------------------------------

plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="Flight_Distance",
    y="Arrival_Delay",
    scatter_kws={"alpha": 0.3}
)

plt.title("Flight Distance vs Arrival Delay")
plt.xlabel("Flight Distance (miles)")
plt.ylabel("Arrival Delay (minutes)")

plt.show()
```

```python
# ------------------------------------------
# 17. Correlation Heatmap
# ------------------------------------------

plt.figure(figsize=(8, 6))

corr = df[
    [
        "Flight_Distance",
        "Departure_Delay",
        "Arrival_Delay"
    ]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
